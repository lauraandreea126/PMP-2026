# Exercitiul 1 - Demonstrarea algoritmului Forward

Un Model Markov Ascuns (HMM) este definit prin:
- Stările ascunse: $X_1, X_2, \dots, X_T$
- Observațiile: $O_1, O_2, \dots, O_T$
- Distribuția inițială a stărilor: $\pi_i = P(X_1 = i)$
- Matricea probabilităților de tranziție: $a_{ij} = P(X_t = j \mid X_{t-1} = i)$
- Probabilitățile de emisie: $b_j(o) = P(O_t = o \mid X_t = j)$

Variabila Forward reprezintă probabilitatea comună de a observa secvența parțială $o_1, o_2, \dots, o_t$ și de a ne afla în starea ascunsă $j$ la momentul $t$:
$$\alpha_t(j) = P(O_1 = o_1, O_2 = o_2, \dots, O_t = o_t, X_t = j)$$

---

## a) Pasul de Inițializare ($t = 1$)

### Enunț
Să se arate că $\alpha_1(j) = \pi_j \cdot b_j(o_1)$.

### Demonstrație pas cu pas
Conform definiției variabilei Forward pentru primul moment de timp ($t = 1$):
$$\alpha_1(j) = P(O_1 = o_1, X_1 = j)$$

Aplicăm **regula produsului** (sau definiția probabilității condiționate):
$$P(A, B) = P(B) \cdot P(A \mid B)$$

În cazul nostru, alegem $A = (O_1 = o_1)$ și $B = (X_1 = j)$:
$$\alpha_1(j) = P(X_1 = j) \cdot P(O_1 = o_1 \mid X_1 = j)$$

Înlocuim termenii conform definițiilor parametrilor modelului HMM:
1. $P(X_1 = j) = \pi_j$ este probabilitatea apriori ca sistemul să pornească în starea $j$.
2. $P(O_1 = o_1 \mid X_1 = j) = b_j(o_1)$ este probabilitatea ca starea $j$ să genereze/emită observația $o_1$.

Așadar, prin înlocuire directă, obținem:
$$\alpha_1(j) = \pi_j \cdot b_j(o_1)$$

---

## b) Pasul de Recurență ($t = 2, \dots, T$)

### Enunț
Să se arate că $\alpha_t(j) = b_j(o_t) \sum_{i} \alpha_{t-1}(i) \cdot a_{ij}$.

### Demonstrație pas cu pas

#### Pasul 1: Marginalizarea stării anterioare (Formula Probabilității Totale)
Pornim de la definiția variabilei Forward la pasul $t$:
$$\alpha_t(j) = P(O_1 = o_1, \dots, O_{t-1} = o_{t-1}, O_t = o_t, X_t = j)$$

Includem starea ascunsă de la momentul anterior $X_{t-1}$ sumând peste toate stările posibile $i$ din spațiul stărilor (conform Formulei Probabilității Totale):
$$\alpha_t(j) = \sum_{i} P(O_1 = o_1, \dots, O_{t-1} = o_{t-1}, O_t = o_t, X_{t-1} = i, X_t = j)$$

#### Pasul 2: Descompunerea probabilității comune prin Regula Produsului
Descompunem termenul din sumă folosind regula lanțului de probabilități:
$$P(A, B, C, D) = P(A) \cdot P(B \mid A) \cdot P(C \mid A, B)$$

Alegem:
- $A = (O_1 = o_1, \dots, O_{t-1} = o_{t-1}, X_{t-1} = i)$ (istoricul până la $t-1$)
- $B = (X_t = j)$ (starea la momentul $t$)
- $C = (O_t = o_t)$ (observația la momentul $t$)

Ecuația devine:
$$\alpha_t(j) = \sum_{i} P(O_1 = o_1, \dots, O_{t-1} = o_{t-1}, X_{t-1} = i) \cdot P(X_t = j \mid X_{t-1} = i, O_1 = o_1, \dots, O_{t-1} = o_{t-1}) \cdot P(O_t = o_t \mid X_t = j, X_{t-1} = i, O_1 = o_1, \dots, O_{t-1} = o_{t-1})$$

#### Pasul 3: Aplicarea ipotezelor fundamentale ale modelului HMM

1. **Definiția variabilei Forward la pasul $t-1$**:
   $$P(O_1 = o_1, \dots, O_{t-1} = o_{t-1}, X_{t-1} = i) = \alpha_{t-1}(i)$$

2. **Proprietatea Markov de ordinul 1 pentru stările ascunse**:
   Starea curentă $X_t$ depinde **doar** de starea imediat anterioară $X_{t-1}$, fiind independentă de tot istoricul observațiilor anterioare:
   $$P(X_t = j \mid X_{t-1} = i, O_1 = o_1, \dots, O_{t-1} = o_{t-1}) = P(X_t = j \mid X_{t-1} = i) = a_{ij}$$

3. **Independența condiționată a emisiilor**:
   Observația curentă $O_t$ depinde **doar** de starea ascunsă curentă $X_t$, fiind independentă de stările și observațiile trecute:
   $$P(O_t = o_t \mid X_t = j, X_{t-1} = i, O_1 = o_1, \dots, O_{t-1} = o_{t-1}) = P(O_t = o_t \mid X_t = j) = b_j(o_t)$$

#### Pasul 4: Asamblarea formulei
Înlocuind cele trei simplificări în sumă, obținem:
$$\alpha_t(j) = \sum_{i} \alpha_{t-1}(i) \cdot a_{ij} \cdot b_j(o_t)$$

Deoarece $b_j(o_t)$ reprezintă probabilitatea de emisie la pasul $t$ pentru starea $j$, el nu depinde de indicele de însumare $i$ (starea anterioară). Prin urmare, îl putem scoate în afara sumei ca factor comun:
$$\alpha_t(j) = b_j(o_t) \sum_{i} \alpha_{t-1}(i) \cdot a_{ij}, \quad \text{pentru } t = 2, \dots, T$$

---

## c) Pasul de Finalizare

### Enunț
Să se arate că $P(O_1 = o_1, \dots, O_T = o_T) = \sum_{j} \alpha_T(j)$.

### Demonstrație pas cu pas
Dorim să calculăm probabilitatea marginală de a obține întreaga secvență de observații de la pasul $1$ la pasul $T$.

Conform **Formulei Probabilității Totale**, însumăm probabilitatea comună a observațiilor și a stării finale peste toate stările ascunse posibile $j$ de la momentul final $T$:
$$P(O_1 = o_1, \dots, O_T = o_T) = \sum_{j} P(O_1 = o_1, \dots, O_T = o_T, X_T = j)$$

Conform definiției variabilei Forward la pasul final $t = T$:
$$\alpha_T(j) = P(O_1 = o_1, \dots, O_T = o_T, X_T = j)$$

Înlocuind $\alpha_T(j)$ în sumă obținem probabilitatea totală a secvenței observate:
$$P(O_1 = o_1, \dots, O_T = o_T) = \sum_{j} \alpha_T(j)$$

# Exercitiul 2 - Algoritmul Viterbi (Deducerea dificultatii testelor)

## a) Definirea modelului si verificarea matricilor

Modelul HMM contine:
- 3 stari ascunse (Dificultatile): D (Dificil), M (Mediu), U (Usor)
- 4 observatii (Notele): FB (Foarte Bine), B (Bine), S (Suficient), NS (Nesatisfacator)
- Secventa de 12 note observate: [FB, FB, B, B, S, FB, NS, B, B, FB, FB, S]

Verificam ca suma pe fiecare rand din matricea de tranzitie A si matricea de emisie B este 1.

## b) Initializarea Viterbi
Variabila delta_t(j) reprezinta probabilitatea comuna maxima a unei secvente de stari care se termina in starea j la momentul t, impreuna cu observatiile de la pasul 1 pana la t.

La primul pas (t = 1):
delta_1(j) = pi_j * b_j(o_1)

## c) Recurenta si rolul tabelului psi
Pentru t = 2, ..., T:
delta_t(j) = b_j(o_t) * max_i (delta_{t-1}(i) * a_{ij})
psi_t(j) = argmax_i (delta_{t-1}(i) * a_{ij})

Rolul tabelului psi_t(j): retine indicele starii precedente i de la pasul t-1 care a produs scorul maxim pentru starea j de la pasul t. Fara acest tabel nu am putea face backtracking pentru a reconstrui traseul cel mai probabil de la final spre inceput.

```python
import numpy as np

# a) definirea modelului
# stari: 0: D (Dificil), 1: M (Mediu), 2: U (Usor)
states = ['D', 'M', 'U']
n_states = len(states)

# mapare observatii note
obs_map = {'FB': 0, 'B': 1, 'S': 2, 'NS': 3}
note_labels = ['FB', 'FB', 'B', 'B', 'S', 'FB', 'NS', 'B', 'B', 'FB', 'FB', 'S']
obs = [obs_map[o] for o in note_labels]
T = len(obs)

# distributia initiala pi
pi = np.array([0.3, 0.4, 0.3])

# matricea de tranzitie A (randuri/coloane: D, M, U)
A = np.array([
    [0.5, 0.3, 0.2],  # de la D
    [0.2, 0.6, 0.2],  # de la M
    [0.1, 0.3, 0.6]   # de la U
])

# matricea de emisie B (randuri: D, M, U; coloane: FB, B, S, NS)
B = np.array([
    [0.1, 0.2, 0.4, 0.3],   # D
    [0.3, 0.4, 0.2, 0.1],   # M
    [0.6, 0.3, 0.08, 0.02]  # U
])

# verificam ca sumele pe randuri sa fie 1
print("suma randuri A:", np.sum(A, axis=1))
print("suma randuri B:", np.sum(B, axis=1))


# b) si c) implementarea viterbi cu doua tabele (delta si psi)
delta = np.zeros((n_states, T))
psi = np.zeros((n_states, T), dtype=int)

# pasul b) initializare la t = 0 (prima notă FB)
delta[:, 0] = pi * B[:, obs[0]]

# pasul c) recurenta
for t in range(1, T):
    for j in range(n_states):
        # calculam produsele delta_{t-1}(i) * a_{ij} pentru toti i
        prod = delta[:, t-1] * A[:, j]
        psi[j, t] = np.argmax(prod)
        delta[j, t] = B[j, obs[t]] * np.max(prod)


# d) afisare delta_1 si delta_2 si verificare manuala
print("\n--- d) verificare intermediara delta_1 si delta_2 ---")
print("delta_1 (dupa prima nota FB):")
for j, st in enumerate(states):
    print(f"  delta_1({st}) = {delta[j, 0]:.6f}")

print("\ndelta_2 (dupa a doua nota FB):")
for j, st in enumerate(states):
    print(f"  delta_2({st}) = {delta[j, 1]:.6f}")

# calcul manual pentru delta_2(M):
# max(delta_1(D)*A(D->M), delta_1(M)*A(M->M), delta_1(U)*A(U->M)) * B(M, FB)
p_D_M = delta[0, 0] * A[0, 1]  # 0.03 * 0.3 = 0.009
p_M_M = delta[1, 0] * A[1, 1]  # 0.12 * 0.6 = 0.072
p_U_M = delta[2, 0] * A[2, 1]  # 0.18 * 0.3 = 0.054
max_prev = max(p_D_M, p_M_M, p_U_M)  # max e 0.072 de la M
verif_manuala_M = max_prev * B[1, obs[1]]  # 0.072 * 0.3 = 0.0216

print(f"\nverificare manuala delta_2(M): max({p_D_M:.4f}, {p_M_M:.4f}, {p_U_M:.4f}) * {B[1, obs[1]]} = {verif_manuala_M:.6f}")


# e) reconstructia traseului (backtracking)
best_path_idx = np.zeros(T, dtype=int)

# alegem starea finala cu scor maxim la t = T-1
best_path_idx[T-1] = np.argmax(delta[:, T-1])
max_score = np.max(delta[:, T-1])

# mergem inapoi din pas in pas folosind psi
for t in range(T-2, -1, -1):
    best_path_idx[t] = psi[best_path_idx[t+1], t+1]

best_path_labels = [states[idx] for idx in best_path_idx]

print("\n--- e) secventa finala si scorul maxim ---")
print("note observate:   ", note_labels)
print("dificultati (x*): ", best_path_labels)
print(f"probabilitatea comuna maxima P(x*, o) = {max_score:.6e}")


# optional: algoritmul forward pentru P(o) si calcul P(x* | o)
alpha = np.zeros((n_states, T))
alpha[:, 0] = pi * B[:, obs[0]]

for t in range(1, T):
    for j in range(n_states):
        alpha[j, t] = B[j, obs[t]] * np.sum(alpha[:, t-1] * A[:, j])

p_obs = np.sum(alpha[:, T-1])
p_cond = max_score / p_obs

print("\n--- optional: probabilitati totale ---")
print(f"P(o) prin Forward = {p_obs:.6e}")
print(f"P(x* | o) = P(x*, o) / P(o) = {p_cond:.6f} ({p_cond * 100:.2f}%)")
```

# f) Interpretare teoretica

###1. De ce Viterbi cauta secventa completa si nu fiecare dificultate separat?
Algoritmul Viterbi gaseste secventa globala optima x* = (x_1, x_2, ..., x_12) care maximizeaza probabilitatea comuna P(x, o).
Daca am alege cea mai probabila stare separat la fiecare pas (adica argmax peste probabilitatile marginale P(X_t | o)), am risca sa obtinem o secventa imposibila in practic, cu tranzitii de probabilitate 0 sau foarte mica intre doua teste consecutive.

###2. Ce reprezinta scorul obtinut la punctul e)?
Scorul max_x P(x, o) este o probabilitate COMUNA (a stărilor si a notelor impreuna), NU o probabilitate conditionata P(x | o).
Pentru a afla cat de probabil este traseul gasit dandu-se notele, se imparte aceasta probabilitate comuna la probabilitatea totala a notelor P(o) obtinuta cu algoritmul Forward:
P(x* | o) = P(x*, o) / P(o)
