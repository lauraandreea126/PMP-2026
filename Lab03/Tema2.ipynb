## Exercițiul 1 – Aruncarea a două zaruri

Se aruncă două zaruri, iar rezultatele sunt notate cu X și Y.

Vom rezolva teoretic fiecare subpunct, apoi vom simula 100.000 de aruncări în Python și vom compara rezultatele obținute cu valorile exacte.


### Spațiul rezultatelor posibile

Fiecare zar poate indica unul dintre numerele 1, 2, 3, 4, 5, 6.

Notăm cu X rezultatul primului zar și cu Y rezultatul celui de-al doilea zar.

Numărul total de rezultate posibile este:

$$|\Omega| = 6 \cdot 6 = 36.$$

Presupunem că zarurile sunt echilibrate. Astfel, toate cele 36 de perechi ordonate sunt echiprobabile.

# Formula probabilității clasice:

$$P(E)=\frac{\text{numărul cazurilor favorabile}}{\text{numărul cazurilor posibile}}.$$

Pentru probabilitățile condiționate :

$$P(E\mid F)=\frac{P(E\cap F)}{P(F)}, \quad P(F)>0.$$




### Subpunctul 1

**Cerință:** Calculați probabilitatea ca suma să fie cel puțin 10, știind că primul zar indică un număr par.

Notăm:

$$A=\{X+Y\geq 10\}$$

$$B=\{X\text{ este par}\}$$

Căutăm probabilitatea condiționată $P(A\mid B)$.

Nr. cazurilor care respectă condiția:

Primul zar trebuie să fie par, deci poate avea valorile 2, 4 sau 6. Al doilea zar poate avea oricare dintre cele 6 valori.

Prin urmare:

$$N_B=3\cdot 6=18.$$

Dacă primul zar este 2, suma nu poate fi cel puțin 10.

Dacă primul zar este 4, singura posibilitate este:

$$(4,6).$$

Dacă primul zar este 6, posibilitățile sunt:

$$(6,4),\ (6,5),\ (6,6).$$

Avem în total 4 cazuri favorabile.

Aplicăm formula probabilității condiționate:

$$P(A\mid B)=\frac{4}{18}=\frac{2}{9}\approx 0.2222.$$

**Răspuns:** Probabilitatea este aproximativ 22,22%.



### Subpunctul 2

Notăm cu C evenimentul „cel puțin un zar indică un număr par”.

Calculăm mai întâi numărul cazurilor în care ambele zaruri sunt impare. Fiecare zar poate avea 3 valori impare, deci:

$$3\cdot 3=9.$$

Nr. cazurilor în care cel puțin un zar este par este:

$$N_C=36-9=27.$$

Suma este cel puțin 10 în următoarele cazuri:

$$(4,6), (5,5), (6,4), (5,6), (6,5), (6,6).$$

Dintre acestea, (5,5) nu respectă condiția, deci avem 5 cazuri favorabile.

Aplicăm formula probabilității condiționate:

$$P(A\mid C)=\frac{5}{27}\approx 0.1852.$$

**Răspuns:** Probabilitatea este aproximativ 18,52%.



### Subpunctul 3

Notăm cu D evenimentul „suma este 8” și cu E evenimentul „ambele zaruri sunt pare”.

Rezultatele pentru care suma este 8 sunt:

$$(2,6), (3,5), (4,4), (5,3), (6,2).$$

Avem 5 cazuri posibile. Dintre acestea, 3 au ambele numere pare:

$$(2,6), (4,4), (6,2).$$

Prin urmare:

$$P(E\mid D)=\frac{3}{5}=0.6.$$

**Răspuns:** Probabilitatea este 60%.



### Subpunctul 4

Verificăm dacă evenimentele A și B sunt independente.

Două evenimente sunt independente dacă:

$$P(A\cap B)=P(A)\cdot P(B).$$

Calculăm probabilitățile:

$$P(A)=\frac{6}{36}=\frac{1}{6}.$$

$$P(B)=\frac{18}{36}=\frac{1}{2}.$$

$$P(A\cap B)=\frac{4}{36}=\frac{1}{9}.$$

Dar:

$$P(A)\cdot P(B)=\frac{1}{6}\cdot\frac{1}{2}=\frac{1}{12}.$$

Deoarece $\frac{1}{9}\neq\frac{1}{12}$, evenimentele A și B nu sunt independente.


```python

import numpy as np

n = 100000

X = np.random.randint(1, 7, n)
Y = np.random.randint(1, 7, n)

A = X + Y >= 10
B = X % 2 == 0
C = (X % 2 == 0) | (Y % 2 == 0)
D = X + Y == 8
E = (X % 2 == 0) & (Y % 2 == 0)

print("Subpunctul 1:", A[B].mean(), "Exact:", 2/9)
print("Subpunctul 2:", A[C].mean(), "Exact:", 5/27)
print("Subpunctul 3:", E[D].mean(), "Exact:", 3/5)

print("P(A):", A.mean())
print("P(B):", B.mean())
print("P(A intersectie B):", (A & B).mean())
print("P(A) * P(B):", A.mean() * B.mean())
```


## Exercițiul 2 – Filtrarea mesajelor spam

Un sistem de e-mail primește 30% mesaje spam. Cuvântul „ofertă” apare în 80% dintre mesajele spam și în 15% dintre mesajele legitime.



### Subpunctul 1

Notăm cu S evenimentul „mesajul este spam” și cu O evenimentul „mesajul conține cuvântul ofertă”.

Avem:

$$P(S)=0.30,\quad P(\overline{S})=0.70$$

$$P(O\mid S)=0.80,\quad P(O\mid \overline{S})=0.15$$

Folosim formula lui Bayes:

$$P(S\mid O)=\frac{P(O\mid S)\cdot P(S)}{P(O)}$$

Calculăm probabilitatea ca un mesaj să conțină cuvântul „ofertă”:

$$P(O)=0.80\cdot0.30+0.15\cdot0.70$$

$$P(O)=0.24+0.105=0.345$$

Prin urmare:

$$P(S\mid O)=\frac{0.80\cdot0.30}{0.345}$$

$$P(S\mid O)=\frac{0.24}{0.345}\approx0.6957$$

**Răspuns:** Probabilitatea ca mesajul să fie spam, știind că include cuvântul „ofertă”, este aproximativ 69,57%.



### Subpunctul 2

Notăm cu G evenimentul „mesajul conține cuvântul gratuit”.

Avem:

$$P(G\mid S)=0.60,\quad P(G\mid\overline{S})=0.10$$

Presupunem că apariția celor două cuvinte este independentă condiționat de tipul mesajului.

Pentru mesajele spam:

$$P(O\cap G\mid S)=0.80\cdot0.60=0.48$$

Pentru mesajele legitime:

$$P(O\cap G\mid\overline{S})=0.15\cdot0.10=0.015$$

Aplicăm formula lui Bayes:

$$P(S\mid O\cap G)=
\frac{P(O\cap G\mid S)\cdot P(S)}
{P(O\cap G\mid S)\cdot P(S)+P(O\cap G\mid\overline{S})\cdot P(\overline{S})}$$

Înlocuim valorile:

$$P(S\mid O\cap G)=
\frac{0.48\cdot0.30}{0.48\cdot0.30+0.015\cdot0.70}$$

$$P(S\mid O\cap G)=\frac{0.144}{0.1545}\approx0.9320$$

**Răspuns:** Probabilitatea ca mesajul să fie spam, dacă include ambele cuvinte, este aproximativ 93,20%.



### Subpunctul 3

Ipoteza independenței condiționate poate fi nerealistă deoarece cele două cuvinte pot apărea împreună în funcție de conținutul mesajului.

De exemplu, mesajele care conțin cuvântul „ofertă” pot conține frecvent și cuvântul „gratuit”. Prin urmare, apariția unui cuvânt poate fi legată de apariția celuilalt, chiar dacă știm că mesajul este spam.



### Subpunctul 4

Propunem următoarea regulă de clasificare:

Dacă probabilitatea condiționată ca mesajul să fie spam este mai mare de 0.5, clasificăm mesajul ca spam. Dacă probabilitatea este mai mică sau egală cu 0.5, clasificăm mesajul ca legitim.

Pragul poate fi modificat în funcție de cât de important este să evităm clasificarea greșită a mesajelor.


```python

# prob din enunt
P_S = 0.30
P_L = 0.70

P_O_S = 0.80
P_O_L = 0.15

P_G_S = 0.60
P_G_L = 0.10

# Subpunctul 1
P_S_O = (P_O_S * P_S) / (P_O_S * P_S + P_O_L * P_L)

print("P(spam | oferta) =", P_S_O)

# Subpunctul 2
P_S_OG = (P_S * P_O_S * P_G_S) / (
    P_S * P_O_S * P_G_S + P_L * P_O_L * P_G_L
)

print("P(spam | oferta si gratuit) =", P_S_OG)

# Subpunctul 4
if P_S_OG > 0.5:
    print("Mesajul este spam")
else:
    print("Mesajul este legitim")
```


## Exercițiul 3 – Paradoxul lui Simpson

O universitate analizează rata de promovare la două discipline, A și B, pentru studenții din două programe de studiu.

Datele sunt:

| Program | Disciplina A | Disciplina B |
|---|---:|---:|
| Programul 1 | 81 din 90 | 19 din 20 |
| Programul 2 | 1 din 10 | 16 din 80 |

Valorile reprezintă numărul studenților promovați din totalul celor evaluați.



### Subpunctul 1

Calculăm rata de promovare pentru fiecare disciplină, separat pentru fiecare program.

**Programul 1**

Disciplina A:

$$\frac{81}{90}=0.9=90\%$$

Disciplina B:

$$\frac{19}{20}=0.95=95\%$$

**Programul 2**

Disciplina A:

$$\frac{1}{10}=0.1=10\%$$

Disciplina B:

$$\frac{16}{80}=0.2=20\%$$



### Subpunctul 2

Comparăm ratele de promovare pentru cele două discipline în fiecare program.

În Programul 1, disciplina B are rata de promovare mai mare:

$$95\%>90\%$$

În Programul 2, disciplina B are tot rata mai mare:

$$20\%>10\%$$

Prin urmare, disciplina B are rata de promovare mai mare în ambele programe.



### Subpunctul 3

Calculăm rata globală de promovare pentru fiecare disciplină.

**Disciplina A**

Numărul total de studenți promovați este 81 + 1 = 82, din 90 + 10 = 100 de studenți evaluați.

$$P(A)=\frac{82}{100}=82\%$$

**Disciplina B**

Numărul total de studenți promovați este 19 + 16 = 35, din 20 + 80 = 100 de studenți evaluați.

$$P(B)=\frac{35}{100}=35\%$$

La nivel global, disciplina A are rata de promovare mai mare: 82% față de 35%.



### Subpunctul 4

În fiecare program, disciplina B are rata de promovare mai mare. Totuși, când adunăm datele, disciplina A are rata globală mai mare.

Acest lucru se întâmplă deoarece numărul studenților evaluați diferă între programe pentru fiecare disciplină. La disciplina A, cei mai mulți studenți sunt în Programul 1, unde rata este mare. La disciplina B, cei mai mulți studenți sunt în Programul 2, unde ratele de promovare sunt mai mici.

Astfel, agregarea datelor poate schimba concluzia comparației. Acest fenomen se numește paradoxul lui Simpson.



### Subpunctul 5

Folosind probabilități condiționate, comparăm rata de promovare pentru fiecare disciplină în funcție de program cu rata globală de promovare.

În fiecare program, disciplina B are rata mai mare. Dar, după agregarea datelor, disciplina A are rata globală mai mare.

Prin urmare, distribuția studenților între programe influențează comparația globală. Acesta este paradoxul lui Simpson.


```python

import pandas as pd

date = pd.DataFrame({
    "Disciplina": ["A", "B", "A", "B"],
    "Promovati": [81, 19, 1, 16],
    "Evaluati": [90, 20, 10, 80]
})

date["Rata"] = date["Promovati"] / date["Evaluati"]

print("Ratele pe fiecare program:")
print(date)

total = date.groupby("Disciplina")[["Promovati", "Evaluati"]].sum()

total["Rata"] = total["Promovati"] / total["Evaluati"]

print("\nRatele globale:")
print(total)
```
