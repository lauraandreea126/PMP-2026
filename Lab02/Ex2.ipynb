### a) Probabilitatea ca persoana să fie bolnava

Notăm cu B = { „persoana este bolnava” }, iar cu P = { „testul este pozitiv” }

Din enunt avem:
P(B) = 0.01 ;
P(P|B) = 0.95 ;
P(P'|B') = 0.90 ;

Deci:
P(P|B') = 1 - 0.90 = 0.10

Dorim să calculăm P(B|P). Folosim formula lui Bayes:

P(B|P) = P(P|B) · P(B) / P(P)

Calculam P(P), folosind formula probabilitații totale:

P(P) = P(P|B) · P(B) + P(P|B') · P(B')

Cum P(B') = 1 - 0.01 = 0.99, obtinem:

P(P) = 0.95 · 0.01 + 0.10 · 0.99 = 0.1085

Prin urmare:

P(B|P) = (0.95 · 0.01) / 0.1085 ≈ 0.0876

Deci daca testul este pozitiv, probabilitatea ca persoana sa fie efectiv bolnava este de aproximativ 8.76%.

### b) Simularea in Python

Simulam un numar mare de persoane si generam pentru fiecare situatia
„bolnava” sau „nu este bolnava”, precum si rezultatul testului.

Folosim probabilitatile din enunt: P(B)=0.01, P(P|B)=0.95 si
P(P|B')=0.10.

Pentru fiecare persoana:
- avem probabilitatea 0.01 sa fie bolnava;
- daca este bolnava, testul este pozitiv cu probabilitatea 0.95;
- daca nu este bolnava, testul este pozitiv cu probabilitatea 0.10.
La final calculam dintre persoanele care au avut test pozitiv, cate sunt bolnave.

```python
import numpy as np

nr_pers = 100000

# generam persoanele bolnave
bolnav = np.random.random(nr_pers) < 0.01

# generam rezultatul testului
test_pozitiv = np.zeros(nr_pers, dtype=bool)

for i in range(nr_pers):
    if bolnav[i]:
        test_pozitiv[i] = np.random.random() < 0.95
    else:
        test_pozitiv[i] = np.random.random() < 0.10

# calculam prob ca persoana sa fie bolnava stiind ca testul este pozitiv
probabilitate_simulata = np.mean(bolnav[test_pozitiv])

print("Prob simulata =", probabilitate_simulata)
print("Prob teoretica = 0.0876")
```

In urma simularii obtinem o valoare apropiata de probabilitatea
calculata teoretic la punctul a), adica aproximativ 0.0876.

Diferenta dintre cele doua valori apare deoarece simularea foloseste
valori aleatoare. Cu cat numarul de persoane simulate este mai mare,
cu atat rezultatul simularii se apropie de valoarea teoretica.

### c) Specificitatea minima

Notam cu s specificitatea testului. Dorim ca probabilitatea ca o persoana
sa fie bolnava, stiind ca testul este pozitiv, sa fie de 50%.

Avem:

P(B) = 0.01 ;
P(P|B) = 0.95 ;
P(B') = 0.99 ;
P(P|B') = 1 - s .

Folosim formula lui Bayes:

P(B|P) = P(P|B) · P(B) / P(P)

Dorim ca P(B|P) = 0.5, deci:

0.5 = (0.95 · 0.01) /
       (0.95 · 0.01 + (1-s) · 0.99)

Rezolvand ecuatia obtinem:

0.5 · (0.0095 + 0.99 · (1-s)) = 0.0095

0.00475 + 0.495 · (1-s) = 0.0095

0.495 · (1-s) = 0.00475

1-s = 0.00475 / 0.495

s ≈ 0.9904

Deci specificitatea minima necesara este de aproximativ 99.04%.

Aceasta valoare inseamna ca testul trebuie sa identifice corect aproximativ
99.04% dintre persoanele care nu sunt bolnave pentru ca probabilitatea ca
un rezultat pozitiv sa indice o persoana bolnava sa ajunga la 50%.

```python
s = 0.9904

probabilitate = (0.95 * 0.01) / (
    0.95 * 0.01 + (1 - s) * 0.99
)

print("Specificitatea =", s)
print("P(B|P) =", probabilitate)
```
